```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv(r'C:\Users\Mayur23\OneDrive\Desktop\pareet\Student Feedback Analysis System\data\feedback_cleaned.csv')
data['Open_feedback'] = data['Open_feedback'].fillna('')

data.isnull().sum()
```

```python
sns.countplot(data = data,x = data['Trainer_effectiveness'])
plt.title('Distribution of Trainer Effectiveness Ratings')
plt.savefig(r'outputs\figures\distribution_of_trainer_effectiveness.png')
plt.show()
```

```python
sns.countplot(data = data,x = data['Curriculum_relevance'])
plt.title('Distribution of Curriculum relevance Ratings')
plt.savefig(r'outputs\figures\distribution_of_curriculum_relevance.png')
plt.show()
```

```python
sns.countplot(data = data,x = data['Pace_of_delivery'])
plt.title('Distribution of Pace of delivery Ratings')
plt.savefig(r'outputs\figures\distribution_of_pace_of_delivery.png')
plt.show()
```

```python
ratings_cols = ['Trainer_effectiveness','Curriculum_relevance','Pace_of_delivery','Support_Mentorship','Platform']
print(data.groupby('Program')[['Pace_of_delivery', 'Platform']].mean().sort_values('Platform'))
```

```python
trend = data.groupby('Batch_date')[['Pace_of_delivery', 'Platform']].mean().sort_values('Platform')
trend.plot(kind = 'line' , marker = 'o')
plt.ylabel('Average Rating')
plt.savefig(r'outputs\figures\trends_of_platform_&_pace_of_delivery.png')
plt.show()
```

```python
data['Would_recommend_num'] = data['Would_recommend'].map({'Yes':1,
                                                          'No':0})
corr_cols = ratings_cols + ['Would_recommend_num','Overall_satisfaction']
sns.heatmap(data[corr_cols].corr(),
            annot = True,
            cmap = 'coolwarm')
plt.savefig(r'outputs\figures\correlation_plot.png')
plt.show()
```
