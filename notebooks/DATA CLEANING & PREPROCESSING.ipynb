```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data_cleaned = pd.read_excel(r'C:\Users\Mayur23\OneDrive\Desktop\pareet\Student Feedback Analysis System\data\feedback_raw.xlsx')
```

```python
data_cleaned.info()
```

```python
data_cleaned.columns = ['Respondent_id','Program','Batch','Institution','Delivery_mode','Gender','Academic_year','Trainer_effectiveness','Curriculum_relevance','Pace_of_delivery','Support_Mentorship','Platform','Overall_satisfaction','Would_recommend','Open_feedback']
```

```python
data_cleaned.sample(5)
```

```python
data_cleaned['Delivery_mode'].unique()
```

```python
Delivery_mode_mapping = {
    'Hybrid' : 'Hybrid',
    'Live Online' : 'Online',
    'Classroom' : 'Offline'
}

data_cleaned['Delivery_mode'] = data_cleaned['Delivery_mode'].map(Delivery_mode_mapping)
```

```python
data_cleaned['Academic_year'].unique()
```

```python
data_cleaned['Institution'].unique()
```

```python
data_cleaned['Batch'].unique()
```

```python
data_cleaned['Batch_date'] = pd.to_datetime(data_cleaned['Batch'].str.replace(' Batch', ''), format='%b %Y')
```

```python
data_cleaned.isnull().sum()
```

```python
rating_cols = ['Trainer_effectiveness','Curriculum_relevance','Pace_of_delivery','Support_Mentorship','Platform']
SI = SimpleImputer(strategy = 'median')
data_cleaned[rating_cols] = SI.fit_transform(data_cleaned[rating_cols])
```

```python
data_cleaned['Open_feedback'] = data_cleaned['Open_feedback'].fillna('')
```

```python
data_cleaned.isnull().sum()
```

```python
data_cleaned.to_csv(r'C:\Users\Mayur23\OneDrive\Desktop\pareet\Student Feedback Analysis System\data\feedback_cleaned.csv', index=False)
```
